```python
def filter_strings(lst, k):
    res = []
    for s in lst:
        if len(s) > k:
            res.append(s)
    return res

print(filter_strings(["apple", "pie", "banana"], 3))
```

```python
def area(shape, d1, d2=0):
    if shape == "circle":
        ans = 3.14 * d1 * d1
        print("Circle Area:", ans)
    elif shape == "rectangle":
        ans = d1 * d2
        print("Rectangle Area:", ans)

area("circle", 5)
area("rectangle", 4, 6)
```

```python
def split_email(email):
    username, domain = email.split('@')
    print(f"Username: {username}, Domain: {domain}")
split_email("support@google.com")
```

```python
def sum_nums(lst):
    total = 0
    for x in lst:
        if type(x) == int or type(x) == float:
            total += x
    return total

print("Total Sum:", sum_nums([10, "apple", 20, True, 5]))
```

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, name):
        self.book_table.append(name)

    def customer_order(self, name, item):
        self.customer_orders.append((name, item))

    def print_menu(self):
        print("Menu:")
        for k, v in self.menu_items.items():
            print(k, ":", v)

    def print_reservations(self):
        print("Reservations:")
        for name in self.book_table:
            print(name)

    def print_orders(self):
        print("Orders:")
        for order in self.customer_orders:
            print(order[0], "ordered", order[1])
```
